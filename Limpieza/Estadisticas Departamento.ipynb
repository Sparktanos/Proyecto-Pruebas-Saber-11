# Estadísticas por departamento

```python
import pyspark.sql.functions as F
```

```python
# Educacion (estadisticas) por departamento https://www.datos.gov.co/Educaci-n/MEN_ESTADISTICAS_EN_EDUCACION_EN_PREESCOLAR-B-SICA/ji8i-4anb/about_data
dpto = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("encoding", "UTF-8") \
    .csv("/Volumes/workspace/pdge(saber-11)/saber11/BRONZE/dpto.csv")
```

```python
dpto.printSchema()
```

```python
# Conteo de nulos
dpto.select([F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in dpto.columns]).show()
```

```python
# Inconsistencias case-sensitive
for c in dpto.columns:
    dpto.select(F.count(F.when(F.col(c) != F.lower(F.col(c)),True))).show()
```

```python
# Conteo de duplicados
total     = dpto.count()
distintos = dpto.distinct().count()

print(f"Total filas:     {total}")
print(f"Filas distintas: {distintos}")
print(f"Duplicados:      {total - distintos}")
```

```python
for c in dpto.columns:
    dpto.groupBy(c).count().show()
```

## Filtrado

```python
nuevo = dpto
```

```python
nuevo = nuevo.filter(F.col("AÑO").between(2015, 2023))
```

```python
cols = ["SEDES_CONECTADAS_A_INTERNET","TAMAÑO_PROMEDIO_DE_GRUPO"]
for c in cols:
    nuevo = nuevo.drop(c)
```
