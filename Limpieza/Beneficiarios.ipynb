```python
# =========================================================
# NORMALIZACIÓN BÁSICA DE STRINGS
# =========================================================
# Objetivo:
# - eliminar espacios extra
# - pasar a mayúsculas
# - homogeneizar símbolos como guiones
# Esto ayuda a evitar categorías duplicadas por formato.
string_cols = [c for c, t in df_01.dtypes if t == "string"]

for c in string_cols:
    df_01 = df_01.withColumn(c,F.upper( F.trim( F.regexp_replace(F.col(c), r"\s+", " ") )))

# Unificar guion largo y guion normal para categorías como ETNIA
for c in string_cols:
    df_01 = df_01.withColumn(c, F.regexp_replace(F.col(c), "–", "-") ) 
```

```python
from pyspark.sql import functions as F

# =========================================================
# LIMPIEZA DE FALTANTES SEMÁNTICOS
# =========================================================

# Normalizamos los valores problemáticos para compararlos de forma segura
valores_missing = ["", "ND", "SIN INFORMACIÓN", "SIN INFORMACION", "NULL", "N/A"]
valores_missing_norm = [v.strip().upper() for v in valores_missing]

def es_missing(col_name, missing_list=valores_missing_norm):
    """
    Evalúa si el valor de una columna corresponde a un faltante semántico.
    Se hace cast a string para evitar errores cuando la columna es numérica.
    También se hace trim + upper para manejar espacios y variaciones.
    """
    return F.upper(F.trim(F.col(col_name).cast("string"))).isin(missing_list)

# Columnas donde ND / SIN INFORMACION sí deben convertirse en null
cols_null_directo = [
    "etnia",
    "puntaje_sisben",
    "estrato",
    "codigo_dep_atencion",
    "nombre_dep_atencion",
    "codigo_mun_atencion",
    "nombre_mun_atencion",
    "rango_edad"
]

for c in cols_null_directo:
    df_01 = df_01.withColumn(
        c,
        F.when(es_missing(c), F.lit(None)).otherwise(F.col(c))
    )

# En las columnas de logros se conserva "NO APLICA"
cols_logros = [
    "educacion_inicial",
    "acceso_educacion",
    "sin_trabajo_infantil",
    "ingreso_sobre_pobreza",
    "seguridad_alimentaria",
    "alfabetizacion",
    "herramientas_digitales",
    "sin_hacinamiento"
]

valores_missing_logros = ["", "SIN INFORMACIÓN", "SIN INFORMACION", "NULL", "N/A"]
valores_missing_logros_norm = [v.strip().upper() for v in valores_missing_logros]

for c in cols_logros:
    df_01 = df_01.withColumn(
        c,
        F.when(
            F.upper(F.trim(F.col(c).cast("string"))).isin(valores_missing_logros_norm),
            F.lit(None)
        ).otherwise(F.col(c))
    )

# Para binarias y educación, sí tratamos ND como faltante
cols_binarias_educacion = [
    "sabe_leer_escribir",
    "estudia_actualmente",
    "ultimo_grado_cursado",
    "hogar_ingresos",
    "hogar_falta_comida"
]

for c in cols_binarias_educacion:
    df_01 = df_01.withColumn(
        c,
        F.when(es_missing(c), F.lit(None)).otherwise(F.col(c))
    )
```

```python
# =========================================================
# NORMALIZACIÓN DE CATEGORÍAS
# =========================================================
# Aquí unificamos categorías fragmentadas que representan
# la misma respuesta semántica.

# ETNIA: consolidar variantes equivalentes
df_01 = df_01.withColumn(
    "etnia",
    F.when(F.col("etnia").isin("AFROCOLOMBIANO - NEGRO", "AFROCOLOMBIANO"), "AFROCOLOMBIANO")
     .when(F.col("etnia").isin("ROM O GITANO", "ROM"), "ROM")
     .otherwise(F.col("etnia"))
)

# PE42: corregir una variante con símbolo sobrante
df_01 = df_01.withColumn(
    "ultimo_grado_cursado",
    F.when(F.col("ultimo_grado_cursado") == "TÉCNICO O TECNÓLOGO°", "TÉCNICO O TECNÓLOGO")
     .otherwise(F.col("ultimo_grado_cursado"))
)

# Estandarización simple de respuestas binarias
for c in ["sabe_leer_escribir", "estudia_actualmente", "hogar_falta_comida"]:
    df_01 = df_01.withColumn(
        c,
        F.when(F.col(c) == "SI", "SI")
         .when(F.col(c) == "NO", "NO")
         .otherwise(F.col(c))
    )
```

```python
from pyspark.sql.types import DoubleType

# =========================================================
# TRATAMIENTO ESPECIAL DE HE20
# =========================================================
# Según el profiling, HE20 mezcla:
# - respuestas binarias (SI / NO)
# - algunos valores numéricos aislados
# Por eso la separamos en tres componentes:
# 1) versión binaria
# 2) valor numérico extraído
# 3) bandera de anomalía

# Binario principal
df_01 = df_01.withColumn(
    "hogar_ingresos_bin",
    F.when(F.col("hogar_ingresos") == "SI", 1)
     .when(F.col("hogar_ingresos") == "NO", 0)
     .otherwise(None)
)

# Si la celda trae un monto, lo extraemos como número
df_01 = df_01.withColumn(
    "hogar_ingresos_valor",
    F.when(
        F.col("hogar_ingresos").rlike(r"^[0-9]+(\.[0-9]+)?$"),
        F.col("hogar_ingresos").cast(DoubleType())
    ).otherwise(None)
)

# Marcamos como anomalía cualquier caso que no sea SI/NO
# o que venga en formato numérico
df_01 = df_01.withColumn(
    "hogar_ingresos_anomalia",
    F.when(
        F.col("hogar_ingresos").isNotNull() &
        (
            F.col("hogar_ingresos").rlike(r"^[0-9]+(\.[0-9]+)?$") |
            (~F.col("hogar_ingresos").isin("SI", "NO"))
        ),
        1
    ).otherwise(0)
)
```

```python
# =========================================================
# BINARIZACIÓN DE RESPUESTAS SI / NO
# =========================================================
# Creamos versiones numéricas para facilitar análisis y modelado.
for c in ["sabe_leer_escribir", "estudia_actualmente", "hogar_falta_comida"]:
    df_01 = df_01.withColumn(
        f"{c}_bin",
        F.when(F.col(c) == "SI", 1)
         .when(F.col(c) == "NO", 0)
         .otherwise(None)
    )
```

```python
from pyspark.sql import functions as F

columnas = df_unidos_clean.columns
resumen = []

for c in columnas:
    col_str = F.upper(F.trim(F.col(c).cast("string")))
    
    fila = df_unidos_clean.select(
        F.lit(c).alias("columna"),
        F.sum(F.when(F.col(c).isNull(), 1).otherwise(0)).alias("nulos"),
        F.sum(F.when(col_str == "", 1).otherwise(0)).alias("vacios"),
        F.sum(F.when(col_str == "ND", 1).otherwise(0)).alias("ND"),
        F.sum(F.when(col_str == "SIN INFORMACIÓN", 1).otherwise(0)).alias("SIN_INFO"),
        F.sum(F.when(col_str == "SIN INFORMACION", 1).otherwise(0)).alias("SIN_INFORMACION"),
        F.sum(F.when(col_str == "NULL", 1).otherwise(0)).alias("NULL_LITERAL"),
        F.sum(F.when(col_str == "N/A", 1).otherwise(0)).alias("NA_LITERAL")
    )
    
    resumen.append(fila)

df_resumen = resumen[0]
for fila in resumen[1:]:
    df_resumen = df_resumen.unionByName(fila)

display(df_resumen.orderBy(F.desc("nulos")))
```
