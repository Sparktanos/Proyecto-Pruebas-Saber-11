```python
import pyspark.sql.functions as F
```

```python
victimas_matr = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("encoding", "UTF-8") \
    .csv("/Volumes/workspace/pdge(saber-11)/saber11/BRONZE/victima_matr.csv")
```

```python
null_count = victimas_matr.select([F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in victimas_matr.columns])
print("Columnas con nulos:")
for c in null_count.columns:
    count = null_count.select(c).collect()[0][0]
    if count > 0:
        print(c, f'({count})')
```

```python
# Conteo de nulos
victimas_matr.select([F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in victimas_matr.columns]).show()
```

```python
for c in victimas_matr.columns:
    victimas_matr.groupBy(c).count().show()
```

```python
# Conteo de duplicados
total     = victimas_matr.count()
distintos = victimas_matr.distinct().count()

print(f"Total filas:     {total}")
print(f"Filas distintas: {distintos}")
print(f"Duplicados:      {total - distintos}")
```
