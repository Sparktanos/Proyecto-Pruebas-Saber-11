```python
transformada = victimas_matr
```

```python

transformada = transformada.withColumn(
    "MUNICIPIO",
    F.trim(
        F.lower(
            F.regexp_replace(F.col("MUNICIPIO"), r"\s+", " ")
        )
    )
)

mpios = mpios.withColumn(
    "MUNICIPIO",
    F.trim(
        F.lower(
            F.regexp_replace(F.col("MUNICIPIO"), r"\s+", " ")
        )
    )
)
```

```python
transformada = transformada.dropna()
```

```python
mpios = mpios.withColumn("MUNICIPIO",F.lower(F.col("MUNICIPIO")))
final = transformada.join(mpios, ["MUNICIPIO"], "inner")
```

```python
final.select([F.count(F.when(F.col(c).isNull(), c)).alias(c) for c in final.columns]).show()
```

```python
final = final.withColumn("Total",F.col("FEMENINO")+F.col("MASCULINO"))
final.display()
```

```python
final = final.withColumnRenamed("CÓDIGO DANE DEL MUNICIPIO","COD_MUNICIPIO" )
```

```python
final.createOrReplaceTempView("victima_matrView")
spark.sql("SELECT COUNT(DISTINCT COD_MUNICIPIO)FROM victima_matrView").show()
```

```python
final.write.mode("overwrite").parquet("/Volumes/workspace/pdge(saber-11)/saber11/SILVER/victima_matr/")
```
