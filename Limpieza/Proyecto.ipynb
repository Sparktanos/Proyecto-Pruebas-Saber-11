# Carga de datos

```python
# Pruebas saber 11 https://www.datos.gov.co/Educaci-n/Resultados-nicos-Saber-11/kgxf-xxbe/about_data
Pruebas = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("encoding", "UTF-8") \
    .csv("/Volumes/workspace/pdge(saber-11)/saber11/BRONZE/saber.csv")
```

```python
# Educacion (estadisticas) por municipio https://www.datos.gov.co/Educaci-n/MEN_ESTADISTICAS_EN_EDUCACION_EN_PREESCOLAR-B-SICA/nudc-7mev/about_data
mpio = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("encoding", "UTF-8") \
    .csv("/Volumes/workspace/pdge(saber-11)/saber11/BRONZE/mpio.csv")
```

```python
# Educacion (estadisticas) por departamento https://www.datos.gov.co/Educaci-n/MEN_ESTADISTICAS_EN_EDUCACION_EN_PREESCOLAR-B-SICA/ji8i-4anb/about_data
dpto = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("encoding", "UTF-8") \
    .csv("/Volumes/workspace/pdge(saber-11)/saber11/BRONZE/dpto.csv")
```

```python
beneficiarios = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("encoding", "UTF-8") \
    .csv("/Volumes/workspace/pdge(saber-11)/saber11/BRONZE/beneficiarios.csv")
```

```python
victimas_matr = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("encoding", "UTF-8") \
    .csv("/Volumes/workspace/pdge(saber-11)/saber11/BRONZE/victima_matr.csv")
```
